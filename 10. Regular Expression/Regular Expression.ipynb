- Regular expressions (regex or regexp) are powerful tools for pattern matching and text manipulation. They allow you to search for, match, and manipulate text based on specific patterns.

## Concepts that are used in regular expression 

### 1. Literal Characters:
A regular expression can be composed of literal characters. For example, the regex `hello` would match the word "hello" in a text.

### 2. Metacharacters:
Metacharacters are characters that have a special meaning in regex. Some common metacharacters are `.`, `^`, `$`, `*`, `+`, `?`, `|`, `()`, `[]`, `{}`.

   - `.`: Matches any character except a newline.
   - `^`: Anchors the match to the start of the line.
   - `$`: Anchors the match to the end of the line.
   - `*`: Matches 0 or more occurrences of the preceding character or group.
   - `+`: Matches 1 or more occurrences of the preceding character or group.
   - `?`: Matches 0 or 1 occurrence of the preceding character or group.
   - `|`: Alternation, matches either the expression before or after the pipe.
   - `()`: Groups expressions together.
   - `[]`: Specifies a character class.
   - `{}`: Specifies the number of occurrences.

### 3. Character Classes:
Character classes are enclosed in square brackets `[]` and allow you to match any one of the characters within the brackets. For example, `[aeiou]` matches any vowel.

### 4. Quantifiers:
Quantifiers specify how many times a character or group should appear. Common quantifiers are `*`, `+`, `?`, and `{}`.

### 5. Anchors:
Anchors specify where a match should occur in the text. `^` anchors the match to the start of the line, and `$` anchors it to the end of the line.

### 6. Escaping:
Use a backslash `\` before a metacharacter to match it as a literal character. For example, `\.` matches a period.
